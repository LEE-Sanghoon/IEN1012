###  문자열 인덱싱

* 인덱스 index 는 시퀀스 자료에서 빠르게 찾을 수 있도록 일정한 규칙을 가지고 순서대로 줄 세워 놓을 때, 정보 위치를 가리키는 순서 값
    ```python
    sequence[index]
    ```

````python
# 다음 코드의 실행 결과를 예측해보자.

a = "Hello, WORLD"

print(a[0])
print(a[10])
print(a[-1])
````

### 문자열 슬라이싱

* 슬라이싱 slicing 은 시퀀스 sequence 자료에서 시작 인덱스부터 끝 인덱스 -1 까지의 자료를 잘라 오는 것(복사)
    ```python
    sequence[start:end:step]
    ```

````python
# 다음 코드의 실행 결과를 예측해보자.

a = "Hello, WORLD"

print(a[0:5])
````

````python
a[-5:-1]
````

````python
a[-1:-5:-1]
````

````python
"Hello, WORLD"[::-1]
````

````python
"Hello, WORLD"[-1::-1]
````

* 다음 코드의 실행결과를 확인하고, 그 이유를 생각하자.
    - 인덱싱은 범위가 벗어나면 즉시 IndexError를 발생
    - 슬라이싱은 유효 범위를 벗어나더라도 에러 대신 존재하는 범위까지만 알아서 필터링 --> 범위 관용성 Overslicing Tolerance

````python
""[-1]      # 시퀀스의 처음과 끝을 동적으로 계산해서 통째로 뒤집어줘.
````

````python
""[-1::-1]  # 인덱스 -1 위치(맨 끝 요소)를 고정 시작점으로 잡고 역순으로 잘라줘.
````

### 문자열 포맷팅

* 문자열 서식(format)을 이용하여 변수의 값을 정해진 위치에 출력한다.
    - `%`와 포맷 문자를 사용하는 방법
    - `{<index>}` 와 `.format()` 메소드를 사용하는 방법

````python
print( "I think %s is the best fruit"%("apple") )
````

````python
print( "I think {0} is the best fruit".format("apple") ) 
````

````python
"%s %s %s" % ("apple", "banana", "orange")
````

````python
"{} {} {}".format("apple", "banana", "orange")
````

````python
"{1} {2} {0}".format("apple", "banana", "orange")
````
