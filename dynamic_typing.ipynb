# 변수

파이썬의 변수는 동적 타이핑 (Dynamic Typing), 이름표 (Reference), 다중 할당, 언패킹 (Unpacking), 자동 메모리 관리 (Garbage Collection)의 특징을 갖고 있다.

* `id()`

    객체의 실제 주소(레퍼런스)를 확인하는 내장함수

```python
x = "Hello, World"
id(x)
```

```python
y = x
id(y)
```

```python
x = 20
print("x' ID = ", id(x))
print("y' ID = ", id(y))
```

```python
x = x + 1
print("x' ID = ", id(x))
```
