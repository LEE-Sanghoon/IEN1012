### 자료형

파이썬의 주요 내장 자료형(Built-in Type)은 다음과 같다.

1. 숫자형 (Numberic)
- 정수형 (**int**) : 소수점이 없는 숫자
- 실수형 (**float**) : 소수점이 있는 숫자
- 복소수형 (**complex**) : 실수와 허수로 이루어진 숫자


2. 문자열 (Sequence/String)
- 문자열 (**str**) : 따옴표 (', ")로 감싼 텍스트 데이터


3. 논리형 (Boolean)
- 불 (**bool**) : 오직 참(True)과 거짓(False) 두 가지 값만 있는 자료형


4. 군집/시퀀스형 (Collection)
- 리스트 (**list**) : 여러 값을 순서대로 저장하여 수정 가능한 가변 (Mutable) 자료형 ( [] 사용 )
- 튜플 (**tuple**) : 여러 값을 순서대로 저장하지만 수정할 수 없는 불변 (Immutable) 자료형 ( () 사용 )
- 딕셔너리 (**dict**) : 키 (KEY)와 값(VALUE)의 쌍으로 이루어진 매핑 자료형 ( {} 사용 )
- 세트/집합 (**set**) : 중복을 허용하지 않고 순서가 없는 자료형 ( {} 사용 )


5. 기타 자료형
- NoneType (None) : 값이 없음을 나타내는 단일 객체

````python
x = input("숫자를 입력하세요: ")
x = int(x)

y = input("숫자를 입력하세요: ")
y = int(y)

print( x, " + ", y, " = ", x + y)
````

**[실습]** 20초를 세어 맞히는 프로그램

다음의 코드를 실행하고, 20초를 마음으로 세어 맞춰 보는 게임을 한다.
실행전에 코드를 한 줄씩 읽고, 동작 방법을 이해한다.

````python
import time

input("엔터를 누르고 20초를 셉니다.")
start = time.time()

input("20초 후에 다시 엔터를 누릅니다.")
end = time.time()

et = end - start    # end 시간에서 start 시간을 빼면 실제 걸린 시간 계산

print("실제 시간: ", et, "초")
print("차이: ", abs(et - 20), "초")
````

**[실습]** 마음대로 걷는 거북이

다음의 코드를 바탕으로 한 번에 움직이는 거리를 5~20 거리가 되도록 프로그램을 수정 작성하시오.

````python
import turtle as t
import random

t.shape("turtle")                   # '거북이' 모양의 거북이 그래픽 사용
t.speed(1)                          # 최고 속도

for x in range(500):                # 거북이를 500번 움직이도록 수행
    a = random.randint(1, 360)      # 1~360에서 임의의 수를 뽑아 a에 저장
    t.setheading(a)                 # 거북이 방향을 a 각도로 회전
    t.forward(10)                   # 거북이가 10만큼 앞으로 이동
````

### `while` 명령으로 반복문 작성하기

- 기본 구조:

    ```python
    while 판단_조건:
        반복_실행_문장
    ```

````python
# 1부터 10까지 출력하는 예시

x = 1
while x <= 10:       # x가 10 이하인 동안에
    print(x)
    x += 1           # x = x + 1
````
