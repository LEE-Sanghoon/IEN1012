# 거북이 대포 게임

* ‘곡사포’ 게임의 한 종류
    - 키보드 방향키 ↑, ↓ 로 발사 각도 조절
    - SpaceBar 로 대포 발사
    - 화살촉 모양의 대포가 하늘로 날아감
    - 대포가 땅에 닿을 때 초록색 목표 지점을 맞히면 ‘Good’ 메시지 출력, 빗나가면 ‘Bad!’ 메시지 출력
    - 발사 각도만 조절 가능하고, 힘은 조절 불가

* 환경 설정

Turtle Graphics를 ipynb 형식으로 실행할 때, 표준PC환경 수행과 구글 코랩 환경의 차이점을 고려해야 한다. 표준 turtle 환경과 코랩 ColabTurtle 환경은 좌표 기준과 표현에 차이가 있다. 이를 개선하고자 본 실습에서는 ColabTurtlePlus를 사용한다.

```python
try:
    import ColabTurtlePlus
except ImportError:
    ! pip3 install ColabTurtlePlus

try:
    import ipywidgets
except ImportError:
    ! pip3 install ipywidgets

import ColabTurtlePlus.Turtle as turtle

from ipywidgets import Button, HBox, VBox, Layout
from IPython.display import display
import random
```

* 곡사포 게임

```python
turtle.clearscreen()

turtle.setup(800, 800)
turtle.bgcolor("white")

t = turtle.Turtle()

def turn_up(btn):      # Up-key를 눌렀을 때 호출되는 함수 
    t.left(2)       # 거북이를 왼쪽으로 2도 돌립니다.
    t.update()

def turn_down(btn):    # Down-key를 눌렀을 때 호출되는 함수
    t.right(2)      # 거북이를 오른쪽으로 2도 돌립니다.
    t.update()

def fire(btn):         # SpaceBar를 누르면 거북이 대포를 발사합니다.
    angle = t.heading() # 현재 거북이가 바라보는 각도를 기억합니다.
    while t.ycor() > 0: # 거북이가 땅 위에 있는 동안 반복합니다.
        t.forward(15)   # 15만큼 앞으로 이동합니다.
        t.right(5)      # 오른쪽으로 5도 회전합니다.

    # while 반복문을 빠져나오면 거북이가 땅에 닿은 상태 입니다.
    d = t.distance(target, 0)       # 현재 위치와 목표지점 사이 거리
    t.sety(random.randint(10, 100)) # 메시지 표시할 위치를 선정

    if d < 25:                      # 거리 오차가 25보다 작으면 명중처리               
        t.color("blue")
        t.write("Good!", align="center")
    else:
        t.color("red")              # 그렇지 않으면 실패한 것으로 처리
        t.write("Bad!", align="center")

    t.color("black")    # 거북이 색을 검은색으로 되돌립니다.
    t.goto(-200, 10)    # 거북이 위치를 처음 발사했던 곳으로 되돌립니다.
    t.setheading(angle) # 각도도 처음 기억해 둔 각도로 되돌립니다.


# 땅을 그립니다.
t.goto(-300, 0)
t.down()
t.goto(300, 0)


# 목표 지점을 설정하고 그립니다.
target = random.randint(50, 150)    # 목표 지점을 50~150 사이에 있는 임의의 수로 지정합니다.
t.pensize(3)
t.color("green")
t.up()
t.goto(target - 25, 2)
t.down()
t.goto(target + 25, 2)


# 거북이 색을 검은색으로 지정하고 처음 발사했던 곳으로 되돌립니다.
t.color("black")
t.up()
t.goto(-200, 10)
t.setheading(20)

# GUI 버튼 레이아웃 설정
btn_layout = Layout(width='150px', height='40px')

btn_up   = Button(description='▲ Up', layout=btn_layout)
btn_down = Button(description='▼ Down', layout=btn_layout)
btn_fire = Button(description='SpaceBar', layout=btn_layout)

control_panel = VBox([
    HBox([btn_up, btn_down, btn_fire], layout=Layout(justify_content='center'))
])

# 거북이가 동작하는 데 필요한 설정을 합니다.
btn_up.on_click(turn_up)
btn_down.on_click(turn_down)
btn_fire.on_click(fire)

display(control_panel)
```
